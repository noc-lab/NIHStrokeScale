```python
import pandas as pd
import hashlib
from datetime import datetime
from pathlib import Path

# File paths
GOLD_PATH = "../postproc_inputs/NIH_GoldAndAutoExtraction_AllInputSettingsNoAE_250806.csv"
SILVER_PATH = "../postproc_inputs/gold_silver.xlsx"
MAPFILE_OUT = "../postproc_intermediate/mapfile.csv"
SEED = "fixed_seed"

# Define NIHSS item labels
ITEMS = ['1a','1b','1c','2','3','4','5a','5b','6a','6b','7','8','9','10','11']
```

```python
# Load gold (CSV) and silver (XLSX) files
gold_df = pd.read_csv(GOLD_PATH)
silver_df = pd.read_excel(SILVER_PATH)

print("Gold rows:", len(gold_df))
print("Silver rows:", len(silver_df))
```

```python
merged_df = pd.merge(
    gold_df,
    silver_df,
    on="Sample_ID",
    suffixes=("_gold", "_silver")
)
print("Merged rows:", len(merged_df))
```

```python
import re

# Define parser
def parse_nihss_breakdown(text, items):

    out = {f'gold_{k}': 0 for k in ITEMS}  # default to 0
    if pd.isna(text):
        return out

    segments = text.split(":")
    for i, label in enumerate(ITEMS):
        if i+1 >= len(segments):
            continue
        right = segments[i+1].strip()
        tokens = right.split()
        if tokens:
            first = tokens[0]
            second = tokens[1] if len(tokens) > 1 else None

        def is_titlelike(tok):
            return bool(re.match(r"^(1[abc]?|2|3|4|5[ab]?|6[ab]?|7|8|9|10|11)$", tok))

        if first.isnumeric() and is_titlelike(second):
            score = int(first)
        elif is_titlelike(second):
            score = 0
        elif i+1 == len(ITEMS) and first.isnumeric():
            score = int(first)
        else:
            score = 0

        out[f'gold_{label}'] = score

    return out
```

```python
# Apply parser function to extract item-level scores
parsed_items = merged_df["NIHSS Breakdown_sc_rerun_breakdown"].apply(lambda x: parse_nihss_breakdown(x, ITEMS)).apply(pd.Series)

# Update merged_df with individual gold item scores
merged_df.update(parsed_items)

# Check for mismatches
merged_df["gold_sum_check"] = parsed_items[[k for k in parsed_items.columns if k.startswith("gold_")]].sum(axis=1)

mismatches = merged_df[merged_df["Manual NIHSS_sc_rerun_score"] != merged_df["gold_sum_check"]]

print(f"Mismatches: {len(mismatches)} / {len(merged_df)}\n")
for _, row in mismatches.iterrows():
    print(f"Sample_ID: {row['Sample_ID']}")
    print(f"Reported Total: {row['Manual NIHSS_sc_rerun_score']}, Parsed Sum: {row['gold_sum_check']}")
    print(f"Breakdown: {row['NIHSS Breakdown_sc_rerun_breakdown']}")
    print("-" * 80)
```

```python
merged_df = merged_df.join(parsed_items, how="left")
merged_df["gold_NIHSS_total"] = merged_df["gold_sum_check"] # use the parsed sum because it is more reliable
merged_df.rename(columns={"NIHSS.y": "silver_NIHSS_total"}, inplace=True) # rename silver total

# Convert admit_date to datetime
merged_df["admit_date"] = pd.to_datetime(merged_df["admit_date"], errors="coerce")

# Create year bin: pre2015 vs post2015
merged_df["year_bin"] = merged_df["admit_date"].dt.year.apply(
    lambda y: "pre2015" if y < 2015 else "post2015"
)

# Map string labels to numeric class IDs
merged_df["dataset_source_flag"] = merged_df["source_silver"].str.lower().map({
    "bmc": 0,
    "mgb": 1,
    "mimic": 2
})
```

```python
# -----------------------
# Build fold assignments from CSV files
# -----------------------

fold_dir = Path("../postproc_inputs/folded_data_4splits_250806_SAMPLE_IDS_ONLY")
fold_files = {
    1: fold_dir / "FOLD1_TEST.csv",
    2: fold_dir / "FOLD2_TEST.csv",
    3: fold_dir / "FOLD3_TEST.csv",
    4: fold_dir / "FOLD4_TEST.csv",
}

fold_map = []
for fold, f in fold_files.items():
    df_fold = pd.read_csv(f)
    # normalize column name in case it isn't exactly "Sample_ID"
    df_fold = df_fold.rename(columns={df_fold.columns[0]: "Sample_ID"})
    df_fold["fold"] = fold
    fold_map.append(df_fold)

fold_assignments = pd.concat(fold_map, ignore_index=True)

# -----------------------
# (3) Merge into base mapfile
# -----------------------

merged_df = merged_df.merge(fold_assignments, on="Sample_ID", how="left")
```

```python
# Add silver_* columns as NA
for k in ITEMS:
    merged_df[f"silver_{k}"] = None

# Select columns for final mapfile
cols = (
    ["Sample_ID", "gold_NIHSS_total"] +
    [f"gold_{k}" for k in ITEMS] +
    ["silver_NIHSS_total"] +
    [f"silver_{k}" for k in ITEMS] +
    ["admit_date", "year_bin", "source_silver", "dataset_source_flag", "fold"]
)

# Create and save mapfile
mapfile_df = merged_df[cols]
mapfile_df.to_csv(MAPFILE_OUT, index=False)
print(f"✅ Saved mapfile to: {MAPFILE_OUT}")
```
