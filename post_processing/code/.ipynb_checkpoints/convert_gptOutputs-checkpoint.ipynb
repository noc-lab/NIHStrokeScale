```python
# -----------------------
# CELL 1: Imports & Paths
# -----------------------

import pandas as pd
from pathlib import Path

# Input (raw GPT fold-level files from Ben)
in_dir = Path("../_fromBen/processed")

# Output (pipeline-ready evals_final_gpt directory)
out_dir = Path("../evals_final_gpt")
out_dir.mkdir(parents=True, exist_ok=True)

print("Input dir:", in_dir.resolve())
print("Output dir:", out_dir.resolve())
```

```python
# -----------------------
# CELL 2: Filename Parsing Helpers
# -----------------------

import re

# Maps for extraction + base prompt labels
extract_map = {
    "Note": "Note",
    "AE_Extraction": "AE_Extraction",
    "GSExtraction": "GS_Extraction",   # no-underscore variant
    "GS_Extraction": "GS_Extraction",  # underscore variant
}

prompt_map = {
    "zs": "ZS_DA",           # zero-shot direct answer
    "zeroShot": "ZS_DA",     # fallback variant
    "cot": "ZS_CoT",         # catch cot, cot_prompt, zs_cot
    "fs": "FS",              # catch fs, fs_prompt
}

def map_filename(fname: str) -> str:
    """
    Convert raw GPT filenames into pipeline-ready evals_final_gpt filenames.
    Handles ExpandedAbbrev/NoExpandedAbbrev, prompt type, few-shot shottedness,
    and special-cases Note_NoExpandedAbbrev → Note__...
    """
    fname_lower = fname.lower()

    # --- Extraction method ---
    extract = None
    for k, v in extract_map.items():
        if k.lower() in fname_lower:
            extract = v
            break
    if extract is None:
        extract = "unknown"

    # --- Prompt type ---
    prompt = "unknown"
    for k, v in prompt_map.items():
        if k in fname_lower:
            prompt = v
            break

    # --- k-shot (if applicable) ---
    m = re.search(r"(\d+)shot", fname_lower)
    if m and prompt.startswith("FS"):
        prompt = f"FS_{m.group(1)}shot"

    # --- Expanded vs NoExpanded ---
    expanded = "ExpandedAbbrev" if "expandedabbrev" in fname_lower else "NoExpandedAbbrev"

    # --- Special case: Note + NoExpandedAbbrev ---
    if extract == "Note" and expanded == "NoExpandedAbbrev":
        return f"Note__{prompt}_AggrPrompt.csv"

    # --- Normal case ---
    return f"{extract}_{expanded}_250806fix_{prompt}_AggrPrompt.csv"

# Quick tests
print(map_filename("NYUGPT_fold1_GSExtraction_ExpandedAbbrev_250806fix_zs.xlsx"))
# -> GS_Extraction_ExpandedAbbrev_250806fix_ZS_DA_AggrPrompt.csv

print(map_filename("NYUGPT_fold1_GS_Extraction_NoExpandedAbbrev_250806fix_fs_prompt_2shot.xlsx"))
# -> GS_Extraction_NoExpandedAbbrev_250806fix_FS_2shot_AggrPrompt.csv

print(map_filename("NYUGPT_fold1_AE_Extraction_ExpandedAbbrev_250806fix_cot_prompt.xlsx"))
# -> AE_Extraction_ExpandedAbbrev_250806fix_ZS_CoT_AggrPrompt.csv

print(map_filename("NYUGPT_fold1_Note_NoExpandedAbbrev_250806fix_zs.xlsx"))
# -> Note__ZS_DA_AggrPrompt.csv
```

```python
# -----------------------
# CELL 2a: Helper to Parse gpt_NIHSS Text
# -----------------------
import re
import numpy as np

def parse_gpt_nihss_text(text):
    """
    Parse the gpt_NIHSS string: extract item-level scores and sum them.
    Returns np.nan if empty or unparsable.
    """
    if not isinstance(text, str) or text.strip().lower() == "empty":
        return np.nan

    # regex to capture lines like "1a (Level of Consciousness):  0"
    pattern = re.compile(r"^\s*\d+[a-z]?\s.*?:\s*([0-9]+)\s*$", re.MULTILINE)

    scores = []
    for match in pattern.finditer(text):
        try:
            scores.append(int(match.group(1)))
        except ValueError:
            continue

    if not scores:
        return np.nan

    total = sum(scores)

    # # Optional QA: check if GPT reported its own total and flag mismatches
    # reported_match = re.search(r"Total NIH Stroke Scale Score:\s*([0-9]+)", text)
    # if reported_match:
    #     reported = int(reported_match.group(1))
    #     if reported != total:
    #         print(f"⚠️ Mismatch in row: reported={reported}, computed={total}")

    return total
```

```python
# -----------------------
# CELL 3: Collect GPT Predictions
# -----------------------

combined = {}  # dict: {out_name: [dataframes]}

# Iterate over each fold directory explicitly
for fold_dir in sorted(in_dir.glob("fold*")):
    if not fold_dir.is_dir():
        continue
    print(f"\n📂 Processing {fold_dir.name}")

    # Extract fold number as int
    m = re.search(r"\d+", fold_dir.name)
    if m:
        fold_num = int(m.group())
    else:
        raise ValueError(f"Could not parse fold number from {fold_dir.name}")

    # Loop over files in this fold
    for f in fold_dir.glob("NYUGPT_*.*"):
        # Read file (Excel or CSV)
        if f.suffix == ".xlsx":
            df = pd.read_excel(f, engine="openpyxl")
        elif f.suffix == ".csv":
            df = pd.read_csv(f)
        else:
            print(f"⚠️ Skipping {f.name} (not .xlsx or .csv)")
            continue

        # Check required columns
        if "Sample_ID" not in df.columns or "gpt_NIHSS" not in df.columns:
            print(f"⚠️ Skipping {f.name} (missing Sample_ID or gpt_NIHSS)")
            continue

        # Standardize to pipeline-ready format: recompute totals from subscores
        df_out = pd.DataFrame({
            "Sample_ID": df["Sample_ID"],
            "NIHSS_pred_total": df["gpt_NIHSS"].apply(parse_gpt_nihss_text)
        })

        # Drop rows with missing predictions
        before = len(df_out)
        df_out = df_out.dropna(subset=["NIHSS_pred_total"])
        after = len(df_out)
        if after < before:
            print(f"⚠️ Dropped {before - after} rows with missing gpt_NIHSS in {f.name}")

        # Convert to int
        df_out["NIHSS_pred_total"] = df_out["NIHSS_pred_total"].round().astype(int)

        # Add fold number as integer
        df_out["fold"] = fold_num

        # Map raw filename → pipeline output filename
        out_name = map_filename(f.name)

        # Add to dictionary
        if out_name not in combined:
            combined[out_name] = []
        combined[out_name].append(df_out)

        # Debug: show extract/prompt detection
        print(f"  ✅ Processed {f.name} → {out_name} "
              f"(rows={len(df_out)}, fold={fold_num})")

# Summary
print("\nSummary:")
for k, v in combined.items():
    total_rows = sum(len(x) for x in v)
    print(f"  {k}: {len(v)} folds, {total_rows} rows")
```

```python
# -----------------------
# CELL 4: Write Consolidated CSVs
# -----------------------

for out_name, dfs in combined.items():
    # Concatenate across folds
    final_df = pd.concat(dfs, ignore_index=True)

    # Write to CSV in evals_final_gpt
    out_path = out_dir / out_name
    final_df.to_csv(out_path, index=False)

    print(f"✅ Wrote {out_name} with {len(final_df)} rows "
          f"and columns {list(final_df.columns)}")
```

```python
# -----------------------
# CELL 5: Sanity Checks
# -----------------------

# Check that we wrote the expected files
written_files = list(out_dir.glob("*.csv"))
print(f"\n📂 Total files written: {len(written_files)}")

for wf in written_files:
    df_check = pd.read_csv(wf)
    print(f"\n{wf.name}: {len(df_check)} rows")
    print("  Columns:", list(df_check.columns))
    print(df_check.head(3))  # peek at first 3 rows

# Optional: check for duplicate Sample_IDs across folds
all_ids = []
for wf in written_files:
    df_check = pd.read_csv(wf)
    all_ids.extend(df_check["Sample_ID"].tolist())

dup_ids = pd.Series(all_ids).duplicated().sum()
print(f"\n🔎 Duplicate Sample_IDs across all files: {dup_ids}")
```
